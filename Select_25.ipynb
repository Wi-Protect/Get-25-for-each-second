```python
import pandas as pd
```

```python
data = pd.read_csv('./csi_data.csv')
```

```python
data.head()
```

```python
groupedData = data.groupby('time')
```

```python
unique_times = data['time'].unique()
```

```python
unique_times
```

```python
groupedData
```

```python
groupedDataSizes = groupedData.size()
```

```python
len(groupedData)
```

```python
len(groupedDataSizes)
```

```python
canGetCount = 0
cannotGetCount = 0
```

```python
newDataFrame = pd.DataFrame(columns=data.columns)
```

```python
selectedRows = []
```

```python
for i in range(len(groupedData)):
  count = groupedDataSizes[i]
  if count > 24:
    group = groupedData.get_group(unique_times[i])
    factor = count / 25
    for j in range(25):
      rowIndex = int(j * factor)
      selectedRows.append(group.iloc[rowIndex])

  else:
    cannotGetCount += 1

```

```python
newDataFrame = pd.DataFrame(selectedRows)
```
